```python
# ================================================================
# ENTERPRISE PILOT ANALYTICS
# ================================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 75)
print("TASK 20 - ENTERPRISE PILOT ANALYTICS")
print("=" * 75)

# ================================================================
# 1. LOAD CLEANED DATASET
# ================================================================

file_path = "/content/Cleaned_Dataset.xlsx"

df = pd.read_excel(file_path)

print("\n1. DATASET LOADED")
print("-" * 75)
print("Rows    :", len(df))
print("Columns :", len(df.columns))

# ================================================================
# 2. BASIC DATASET INFORMATION
# ================================================================

print("\n2. DATASET INFORMATION")
print("-" * 75)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Shape:")
print(df.shape)

# ================================================================
# 3. MISSING VALUE ANALYSIS
# ================================================================

print("\n3. MISSING VALUE ANALYSIS")
print("-" * 75)

missing = df.isnull().sum()

missing_report = pd.DataFrame({
    "Column": missing.index,
    "Missing_Values": missing.values,
    "Missing_Percentage": (
        missing.values / len(df) * 100
    ).round(2)
})

print(
    missing_report[
        missing_report["Missing_Values"] > 0
    ].to_string(index=False)
)

if missing.sum() == 0:
    print("No missing values found.")

# ================================================================
# 4. DUPLICATE ANALYSIS
# ================================================================

print("\n4. DUPLICATE ANALYSIS")
print("-" * 75)

duplicate_count = df.duplicated().sum()

print("Duplicate Records:", duplicate_count)

if duplicate_count == 0:
    print("Dataset contains no duplicate records.")

# ================================================================
# 5. DESCRIPTIVE STATISTICS
# ================================================================

print("\n5. DESCRIPTIVE STATISTICS")
print("-" * 75)

numeric_df = df.select_dtypes(
    include=np.number
)

descriptive_stats = numeric_df.describe().T

descriptive_stats["median"] = numeric_df.median()
descriptive_stats["variance"] = numeric_df.var()
descriptive_stats["skewness"] = numeric_df.skew()

print(
    descriptive_stats.round(2).to_string()
)

# ================================================================
# 6. KEY BUSINESS KPIs
# ================================================================

print("\n6. KEY BUSINESS KPIs")
print("-" * 75)

total_records = len(df)

total_requests = (
    df["requests"].sum()
    if "requests" in df.columns else np.nan
)

total_completed = (
    df["completed_trips"].sum()
    if "completed_trips" in df.columns else np.nan
)

total_cancelled = (
    df["cancelled_trips"].sum()
    if "cancelled_trips" in df.columns else np.nan
)

total_revenue = (
    df["revenue_inr"].sum()
    if "revenue_inr" in df.columns else np.nan
)

total_cost = (
    df["cost_inr"].sum()
    if "cost_inr" in df.columns else np.nan
)

total_profit = (
    total_revenue - total_cost
    if pd.notna(total_revenue) and pd.notna(total_cost)
    else np.nan
)

average_response_time = (
    df["response_time_min"].mean()
    if "response_time_min" in df.columns else np.nan
)

average_success_rate = (
    df["success_rate_pct"].mean()
    if "success_rate_pct" in df.columns else np.nan
)

average_adoption_rate = (
    df["adoption_rate_pct"].mean()
    if "adoption_rate_pct" in df.columns else np.nan
)

average_satisfaction = (
    df["satisfaction_score"].mean()
    if "satisfaction_score" in df.columns else np.nan
)

overall_success_rate = (
    total_completed / total_requests * 100
    if total_requests > 0 else np.nan
)

cancellation_rate = (
    total_cancelled / total_requests * 100
    if total_requests > 0 else np.nan
)

kpi_table = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Requests",
        "Total Completed Trips",
        "Total Cancelled Trips",
        "Total Revenue (INR)",
        "Total Cost (INR)",
        "Total Profit (INR)",
        "Overall Success Rate (%)",
        "Cancellation Rate (%)",
        "Average Response Time (Min)",
        "Average Adoption Rate (%)",
        "Average Satisfaction Score"
    ],

    "Value": [
        total_records,
        total_requests,
        total_completed,
        total_cancelled,
        total_revenue,
        total_cost,
        total_profit,
        overall_success_rate,
        cancellation_rate,
        average_response_time,
        average_adoption_rate,
        average_satisfaction
    ]
})

print(kpi_table.round(2).to_string(index=False))

# ================================================================
# 7. CATEGORY-WISE ANALYSIS
# ================================================================

print("\n7. CATEGORY-WISE ANALYSIS")
print("-" * 75)

categorical_columns = [
    "city",
    "side",
    "channel",
    "pilot_group",
    "event_type"
]

for col in categorical_columns:

    if col in df.columns:

        print(f"\n--- {col.upper()} ---")

        category_summary = (
            df.groupby(col)
            .agg(
                Records=(col, "count"),
                Requests=("requests", "sum"),
                Completed_Trips=("completed_trips", "sum"),
                Revenue_INR=("revenue_inr", "sum"),
                Avg_Response_Time=("response_time_min", "mean"),
                Avg_Success_Rate=("success_rate_pct", "mean"),
                Avg_Adoption_Rate=("adoption_rate_pct", "mean"),
                Avg_Satisfaction=("satisfaction_score", "mean")
            )
            .reset_index()
        )

        category_summary["Completion_Rate_%"] = (
            category_summary["Completed_Trips"] /
            category_summary["Requests"] * 100
        ).round(2)

        print(
            category_summary.round(2).to_string(
                index=False
            )
        )

# ================================================================
# 8. PILOT VS CONTROL ANALYSIS
# ================================================================

print("\n8. PILOT VS CONTROL ANALYSIS")
print("-" * 75)

if "pilot_group" in df.columns:

    pilot_analysis = (
        df.groupby("pilot_group")
        .agg(
            Records=("pilot_group", "count"),
            Requests=("requests", "sum"),
            Completed_Trips=("completed_trips", "sum"),
            Cancelled_Trips=("cancelled_trips", "sum"),
            Revenue_INR=("revenue_inr", "sum"),
            Cost_INR=("cost_inr", "sum"),
            Avg_Response_Time=("response_time_min", "mean"),
            Avg_Success_Rate=("success_rate_pct", "mean"),
            Avg_Adoption_Rate=("adoption_rate_pct", "mean"),
            Avg_Satisfaction=("satisfaction_score", "mean")
        )
        .reset_index()
    )

    pilot_analysis["Completion_Rate_%"] = (
        pilot_analysis["Completed_Trips"] /
        pilot_analysis["Requests"] * 100
    )

    pilot_analysis["Cancellation_Rate_%"] = (
        pilot_analysis["Cancelled_Trips"] /
        pilot_analysis["Requests"] * 100
    )

    pilot_analysis["Profit_INR"] = (
        pilot_analysis["Revenue_INR"] -
        pilot_analysis["Cost_INR"]
    )

    print(
        pilot_analysis.round(2).to_string(
            index=False
        )
    )

# ================================================================
# 9. PILOT IMPROVEMENT ANALYSIS
# ================================================================

print("\n9. PILOT IMPROVEMENT ANALYSIS")
print("-" * 75)

if (
    "pilot_group" in df.columns and
    df["pilot_group"].nunique() >= 2
):

    pilot = df[
        df["pilot_group"].str.lower() == "pilot"
    ]

    control = df[
        df["pilot_group"].str.lower() == "control"
    ]

    if len(pilot) > 0 and len(control) > 0:

        metrics = [
            "success_rate_pct",
            "adoption_rate_pct",
            "response_time_min",
            "satisfaction_score",
            "revenue_inr",
            "cost_inr"
        ]

        comparison = []

        for metric in metrics:

            pilot_value = pilot[metric].mean()
            control_value = control[metric].mean()

            difference = pilot_value - control_value

            if control_value != 0:
                percentage_change = (
                    difference /
                    abs(control_value) * 100
                )
            else:
                percentage_change = np.nan

            comparison.append([
                metric,
                pilot_value,
                control_value,
                difference,
                percentage_change
            ])

        pilot_comparison = pd.DataFrame(
            comparison,
            columns=[
                "Metric",
                "Pilot_Avg",
                "Control_Avg",
                "Difference",
                "Change_%"
            ]
        )

        print(
            pilot_comparison.round(2).to_string(
                index=False
            )
        )

# ================================================================
# 10. CITY PERFORMANCE ANALYSIS
# ================================================================

print("\n10. CITY PERFORMANCE ANALYSIS")
print("-" * 75)

if "city" in df.columns:

    city_analysis = (
        df.groupby("city")
        .agg(
            Records=("city", "count"),
            Requests=("requests", "sum"),
            Completed=("completed_trips", "sum"),
            Cancelled=("cancelled_trips", "sum"),
            Revenue=("revenue_inr", "sum"),
            Cost=("cost_inr", "sum"),
            Avg_Response=("response_time_min", "mean"),
            Avg_Adoption=("adoption_rate_pct", "mean"),
            Avg_Satisfaction=("satisfaction_score", "mean")
        )
        .reset_index()
    )

    city_analysis["Completion_Rate_%"] = (
        city_analysis["Completed"] /
        city_analysis["Requests"] * 100
    )

    city_analysis["Cancellation_Rate_%"] = (
        city_analysis["Cancelled"] /
        city_analysis["Requests"] * 100
    )

    city_analysis["Profit"] = (
        city_analysis["Revenue"] -
        city_analysis["Cost"]
    )

    print(
        city_analysis.round(2).to_string(
            index=False
        )
    )

# ================================================================
# 11. EVENT TYPE ANALYSIS
# ================================================================

print("\n11. EVENT TYPE ANALYSIS")
print("-" * 75)

if "event_type" in df.columns:

    event_analysis = (
        df.groupby("event_type")
        .agg(
            Records=("event_type", "count"),
            Avg_Requests=("requests", "mean"),
            Avg_Completed=("completed_trips", "mean"),
            Avg_Response_Time=("response_time_min", "mean"),
            Avg_Success_Rate=("success_rate_pct", "mean"),
            Avg_Adoption_Rate=("adoption_rate_pct", "mean"),
            Avg_Satisfaction=("satisfaction_score", "mean")
        )
        .reset_index()
    )

    print(
        event_analysis.round(2).to_string(
            index=False
        )
    )

# ================================================================
# 12. CHANNEL ANALYSIS
# ================================================================

print("\n12. CHANNEL PERFORMANCE")
print("-" * 75)

if "channel" in df.columns:

    channel_analysis = (
        df.groupby("channel")
        .agg(
            Records=("channel", "count"),
            Requests=("requests", "sum"),
            Completed=("completed_trips", "sum"),
            Revenue=("revenue_inr", "sum"),
            Avg_Response=("response_time_min", "mean"),
            Avg_Success=("success_rate_pct", "mean"),
            Avg_Adoption=("adoption_rate_pct", "mean"),
            Avg_Satisfaction=("satisfaction_score", "mean")
        )
        .reset_index()
    )

    channel_analysis["Completion_Rate_%"] = (
        channel_analysis["Completed"] /
        channel_analysis["Requests"] * 100
    )

    print(
        channel_analysis.round(2).to_string(
            index=False
        )
    )

# ================================================================
# 13. SIDE-WISE ANALYSIS
# ================================================================

print("\n13. RIDER VS DRIVER ANALYSIS")
print("-" * 75)

if "side" in df.columns:

    side_analysis = (
        df.groupby("side")
        .agg(
            Records=("side", "count"),
            Requests=("requests", "sum"),
            Completed=("completed_trips", "sum"),
            Revenue=("revenue_inr", "sum"),
            Avg_Response=("response_time_min", "mean"),
            Avg_Success=("success_rate_pct", "mean"),
            Avg_Adoption=("adoption_rate_pct", "mean"),
            Avg_Satisfaction=("satisfaction_score", "mean")
        )
        .reset_index()
    )

    print(
        side_analysis.round(2).to_string(
            index=False
        )
    )

# ================================================================
# 14. CORRELATION ANALYSIS
# ================================================================

print("\n14. CORRELATION ANALYSIS")
print("-" * 75)

correlation_matrix = numeric_df.corr()

print(
    correlation_matrix.round(2).to_string()
)

# Find strongest relationships
corr_pairs = (
    correlation_matrix
    .where(
        np.triu(
            np.ones(correlation_matrix.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .reset_index()
)

corr_pairs.columns = [
    "Variable_1",
    "Variable_2",
    "Correlation"
]

corr_pairs["Absolute_Correlation"] = (
    corr_pairs["Correlation"].abs()
)

corr_pairs = corr_pairs.sort_values(
    "Absolute_Correlation",
    ascending=False
)

print("\nStrongest Relationships:")
print(
    corr_pairs.head(10).round(3).to_string(
        index=False
    )
)

# ================================================================
# 15. OUTLIER ANALYSIS USING IQR
# ================================================================

print("\n15. OUTLIER ANALYSIS")
print("-" * 75)

outlier_results = []

for col in numeric_df.columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    outlier_results.append([
        col,
        Q1,
        Q3,
        IQR,
        lower_bound,
        upper_bound,
        outliers
    ])

outlier_report = pd.DataFrame(
    outlier_results,
    columns=[
        "Column",
        "Q1",
        "Q3",
        "IQR",
        "Lower_Bound",
        "Upper_Bound",
        "Outlier_Count"
    ]
)

print(
    outlier_report.round(2).to_string(
        index=False
    )
)

# ================================================================
# 16. PROFITABILITY ANALYSIS
# ================================================================

print("\n16. PROFITABILITY ANALYSIS")
print("-" * 75)

if all(
    col in df.columns
    for col in ["revenue_inr", "cost_inr"]
):

    df["profit_inr"] = (
        df["revenue_inr"] -
        df["cost_inr"]
    )

    df["profit_margin_pct"] = np.where(
        df["revenue_inr"] > 0,
        df["profit_inr"] /
        df["revenue_inr"] * 100,
        0
    )

    profitability = pd.DataFrame({
        "Metric": [
            "Total Revenue",
            "Total Cost",
            "Total Profit",
            "Average Profit",
            "Average Profit Margin (%)",
            "Minimum Profit",
            "Maximum Profit"
        ],

        "Value": [
            df["revenue_inr"].sum(),
            df["cost_inr"].sum(),
            df["profit_inr"].sum(),
            df["profit_inr"].mean(),
            df["profit_margin_pct"].mean(),
            df["profit_inr"].min(),
            df["profit_inr"].max()
        ]
    })

    print(
        profitability.round(2).to_string(
            index=False
        )
    )

# ================================================================
# 17. DAILY PERFORMANCE ANALYSIS
# ================================================================

print("\n17. DAILY PERFORMANCE ANALYSIS")
print("-" * 75)

if "event_date" in df.columns:

    daily_analysis = (
        df.groupby("event_date")
        .agg(
            Records=("event_date", "count"),
            Requests=("requests", "sum"),
            Completed=("completed_trips", "sum"),
            Cancelled=("cancelled_trips", "sum"),
            Revenue=("revenue_inr", "sum"),
            Cost=("cost_inr", "sum"),
            Avg_Response=("response_time_min", "mean"),
            Avg_Success=("success_rate_pct", "mean"),
            Avg_Adoption=("adoption_rate_pct", "mean"),
            Avg_Satisfaction=("satisfaction_score", "mean")
        )
        .reset_index()
    )

    daily_analysis["Completion_Rate_%"] = (
        daily_analysis["Completed"] /
        daily_analysis["Requests"] * 100
    )

    print(
        daily_analysis.head(15).round(2).to_string(
            index=False
        )
    )

# ================================================================
# 18. SUCCESS CRITERIA CHECK
# ================================================================

print("\n18. PILOT SUCCESS CRITERIA CHECK")
print("-" * 75)

criteria = []

# Example pre-agreed thresholds for analytical evaluation.
# These are clearly defined assumptions for the dry-run and
# should be replaced by actual business-approved thresholds
# when production data is available.

criteria.append([
    "Overall Success Rate",
    overall_success_rate,
    70,
    "PASS" if overall_success_rate >= 70 else "FAIL"
])

criteria.append([
    "Average Adoption Rate",
    average_adoption_rate,
    60,
    "PASS" if average_adoption_rate >= 60 else "FAIL"
])

criteria.append([
    "Average Satisfaction",
    average_satisfaction,
    3.5,
    "PASS" if average_satisfaction >= 3.5 else "FAIL"
])

criteria.append([
    "Average Response Time",
    average_response_time,
    10,
    "PASS" if average_response_time <= 10 else "FAIL"
])

success_criteria = pd.DataFrame(
    criteria,
    columns=[
        "Metric",
        "Actual_Value",
        "Target",
        "Result"
    ]
)

print(
    success_criteria.round(2).to_string(
        index=False
    )
)

# ================================================================
# 19. TOP / BOTTOM PERFORMERS
# ================================================================

print("\n19. PERFORMANCE RANKING")
print("-" * 75)

if "city" in df.columns:

    ranking = city_analysis.copy()

    print("\nTop Cities by Completion Rate:")
    print(
        ranking.sort_values(
            "Completion_Rate_%",
            ascending=False
        )
        .head(5)
        [["city", "Completion_Rate_%"]]
        .round(2)
        .to_string(index=False)
    )

    print("\nTop Cities by Revenue:")
    print(
        ranking.sort_values(
            "Revenue",
            ascending=False
        )
        .head(5)
        [["city", "Revenue"]]
        .round(2)
        .to_string(index=False)
    )

    print("\nTop Cities by Satisfaction:")
    print(
        ranking.sort_values(
            "Avg_Satisfaction",
            ascending=False
        )
        .head(5)
        [["city", "Avg_Satisfaction"]]
        .round(2)
        .to_string(index=False)
    )

# ================================================================
# 20. FINAL ANALYTICAL SUMMARY
# ================================================================

print("\n")
print("=" * 75)
print("FINAL ANALYTICAL SUMMARY")
print("=" * 75)

print(f"""
Total Records              : {total_records:,}
Total Requests              : {total_requests:,.0f}
Total Completed Trips       : {total_completed:,.0f}
Total Cancelled Trips       : {total_cancelled:,.0f}
Total Revenue (INR)         : {total_revenue:,.2f}
Total Cost (INR)            : {total_cost:,.2f}
Total Profit (INR)          : {total_profit:,.2f}
Overall Success Rate        : {overall_success_rate:.2f}%
Cancellation Rate           : {cancellation_rate:.2f}%
Average Response Time       : {average_response_time:.2f} minutes
Average Adoption Rate       : {average_adoption_rate:.2f}%
Average Satisfaction        : {average_satisfaction:.2f}/5
Duplicate Records           : {duplicate_count}
Missing Values              : {int(missing.sum())}
""")

# ================================================================
# 21. EXPORT ANALYSIS RESULTS
# ================================================================

output_file = "/content/Python_Analysis_Report.xlsx"

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    kpi_table.to_excel(
        writer,
        sheet_name="KPI_Summary",
        index=False
    )

    descriptive_stats.to_excel(
        writer,
        sheet_name="Descriptive_Stats"
    )

    missing_report.to_excel(
        writer,
        sheet_name="Missing_Values",
        index=False
    )

    outlier_report.to_excel(
        writer,
        sheet_name="Outlier_Analysis",
        index=False
    )

    success_criteria.to_excel(
        writer,
        sheet_name="Success_Criteria",
        index=False
    )

    corr_pairs.to_excel(
        writer,
        sheet_name="Correlations",
        index=False
    )

    if "pilot_group" in df.columns:
        pilot_analysis.to_excel(
            writer,
            sheet_name="Pilot_Control",
            index=False
        )

    if "city" in df.columns:
        city_analysis.to_excel(
            writer,
            sheet_name="City_Analysis",
            index=False
        )

    if "channel" in df.columns:
        channel_analysis.to_excel(
            writer,
            sheet_name="Channel_Analysis",
            index=False
        )

    if "side" in df.columns:
        side_analysis.to_excel(
            writer,
            sheet_name="Side_Analysis",
            index=False
        )

    if "event_type" in df.columns:
        event_analysis.to_excel(
            writer,
            sheet_name="Event_Analysis",
            index=False
        )

print("\n")
print("=" * 75)
print("PYTHON ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 75)

print("\nAnalysis report created:")
print("/content/Python_Analysis_Report.xlsx")

print("\nIMPORTANT:")
print("No charts or dashboards were created.")
print("Charts and Dashboard will be created in Tableau.")
print("=" * 75)
```
